# Retail Sales Analysis Insights

## Question 1: What is the total revenue generated?

The dataset generated a total revenue of **R324,251.00**.

This means the business earned more than three hundred thousand rand from sales transactions across all stores. Revenue was calculated by multiplying the quantity sold by the unit price for each transaction.

---

## Question 2: Which product categories generate the most revenue?

The category revenue analysis showed that some product categories contribute significantly more revenue than others.

Categories that generate higher revenue are likely to have stronger customer demand and should be prioritised for stock availability, promotions, and marketing campaigns. Lower-performing categories may require further investigation to understand customer purchasing behaviour.

---

## Question 3: Which stores perform best?

The store revenue analysis showed differences in performance across store locations.

Sandton generated the highest revenue, while Rosebank generated the lowest revenue.

This suggests that customer demand and sales performance vary by location. The business can study successful stores and apply similar strategies to improve lower-performing locations.

---

## Question 4: What are the best-selling products?

The product quantity analysis identified the products with the highest sales volumes.

Best-selling products are important because they contribute significantly to revenue and customer satisfaction. These products should remain well stocked to avoid lost sales opportunities.

---

## Question 5: What business recommendations can be made?

Based on the analysis, the following recommendations can be made:

- Increase inventory levels for best-selling products.
- Focus marketing efforts on high-performing categories.
- Investigate the reasons behind lower revenue in certain stores.
- Use sales data to improve inventory planning and forecasting.
- Monitor store and category performance regularly using dashboards and reports.

These recommendations can help improve profitability, customer satisfaction, and operational decision-making.

---

# Overall Conclusion

The retail sales analysis successfully identified revenue trends, store performance, category performance, and top-selling products. The results demonstrate how data analysis can be used to support business decisions and improve operational efficiency. By focusing on high-performing products and stores while addressing weaker areas, the business can increase revenue and improve overall performance.
